```python
corpus= """Learning is one of the most important parts of human life because it allows us to grow, adapt, and understand the world around us. It is not limited to classrooms or textbooks; people learn through experiences, mistakes, conversations, and challenges. In today’s rapidly changing world, continuous learning has become even more important because technology and society are constantly evolving. A person who is willing to learn from failure can turn mistakes into valuable lessons and become stronger over time. Just like a character in an anime who starts weak but becomes powerful through training, real-life growth also comes from patience, consistency, and effort. Ultimately, learning is not about knowing everything, but about remaining curious enough to keep improving.
 """
```

```python
from nltk.stem import PorterStemmer
from nltk.corpus import stopwords
```

```python
import nltk
nltk.download('stopwords')
```

```python
stopwords.words('english')
```

```python
stemmer = PorterStemmer()
sentences = nltk.sent_tokenize(corpus)
```

```python
for i in range(len(sentences)):
    words=nltk.word_tokenize(sentences[i])  
    words=[stemmer.stem(word) for word in words if word not in set(stopwords.words('english'))]
    sentences[i] = ' '.join(words) #converting all the words in sentences
```
