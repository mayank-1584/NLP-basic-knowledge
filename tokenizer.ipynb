```python
corpus = """Hlo my name is mayank sharma. am 21 year old and having my btech in computer science , in delhi technical. campus! """
```

```python
print(corpus)
```

```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')  # newer NLTK versions split this out separately
```

```python
from nltk.tokenize import sent_tokenize
```

```python
documnets = sent_tokenize(corpus)
```

```python
from nltk.tokenize import word_tokenize
word_tokenize(corpus)
```

```python
for sentence in documnets:
    print(word_tokenize(sentence))
```

```python
word = ['help' , 'helper' , 'helping' , 'come' , 'comming' , 'going']
```

```python
from nltk.stem import PorterStemmer
stemming = PorterStemmer()
for words in word:
    print(words+ "-->"+ stemming.stem(words))
```

```python
from nltk.stem import RegexpStemmer
rex = RegexpStemmer(
    'ing$|s$|e$|able$|er$|ming$', min=4
)
for words in word:
    print(words+ "-->"+ rex.stem(words))
```
